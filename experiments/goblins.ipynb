```python
# libraries
import numpy as np # Array manipulation
import librosa # Audio processing
import librosa.display # Audio visualization
import matplotlib.pyplot as plt # Plotting
import IPython.display as ipd # Interactive display of audio in jupyter
from scipy.io import wavfile # Audio file I/O
import pandas as pd # Data manipulation
import soundfile as sf # Better audio file I/O
print("Imports successfull")
```

```python
# Load sample audio from librosa
audio_path = librosa.example("trumpet")
y, sr = librosa.load(audio_path)
print(f"Download complete at {audio_path}")
```

```python
# Display properties of the audio file

# Custom approaches
# test_period = len(y)/sr
# print(test_period)

# Standard approaches
duration = librosa.get_duration(y=y, sr=sr)
print(f"Duration: {duration}")
print(f"Sample rate: {sr}")
print(f"Number of samples: {len(y)}")
print(f"Array shape: {y.shape}")
```

```python
# Play the audio
ipd.display(ipd.Audio(y,rate=sr))
```

```python
# Visualize the wave form
plt.figure(figsize=(14,5))
librosa.display.waveshow(y,sr=sr)
plt.title("Trumpet Wave form")
plt.xlabel("time")
plt.ylabel("amplitude")
plt.show()
```

```python
# Use Short time fourier transform
D = librosa.stft(y)
S_db = librosa.amplitude_to_db(np.abs(D), ref=np.max) # Convert to db scale

plt.figure(figsize=(14,5))
librosa.display.specshow(S_db, sr=sr, x_axis="time", y_axis="hz")
plt.colorbar(format='%+0.2f dB')
plt.title("Spectogram")
plt.show()
           
```

```python
# Compute mel spectogram
S = librosa.feature.melspectrogram(y=y,sr=sr,n_mels=128)
S_db_mel = librosa.power_to_db(S,ref=np.max)

# Display the mel spectogram
plt.figure(figsize=(14,5))
librosa.display.specshow(S_db_mel,sr=sr,x_axis="time", y_axis="hz")
plt.title("Mel spectogram")
plt.colorbar(format='%+2.0f dB')
plt.show()

```
