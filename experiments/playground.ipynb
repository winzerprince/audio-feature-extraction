# Let's try out stuff

```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import librosa.display
import IPython.display as ipd
print("Imports successfull")
```

```python
# A simple sine wave
time = np.linspace(0,1,1000)
amplitude = np.sin(2*np.pi*5*time)
plt.figure(figsize=(14,5))
plt.plot(time,amplitude)
plt.xlabel("time")
plt.ylabel("amplitude")
plt.title("Sine wave")
plt.show()
```

```python
# Listen to the sine waves over a duration
d = 5 # Duration in seconds
sr = 22050 # Sample rate
f = 440 # Frequency
pts = 200 # number of array elements to plot

x = np.linspace(0,d, sr*d) # Array of length equal to number of samples in 5 second audio
y = np.sin(2*np.pi*f*x) # Array of amplitudes at each time frame

plt.figure(figsize=(14,5))
plt.xlabel("time")
plt.ylabel("amplitude")
plt.title(f"Waveform of frequency {f}")
plt.plot(x[:pts],y[:pts]) # Plot a small zoomed-in section of the sine wave
plt.show()

ipd.Audio(y,rate=sr)


```

```python
# 2 sine waves combined to make a composite signal
d1 = 5 # Duration in seconds
sr1 = 22050 # Sample rate
f1 = 200 # Frequency
pts = 200 # number of array elements to plot

x1 = np.linspace(0,d1, sr1*d1) # Array of length equal to number of samples in 5 second audio
y1 = np.sin(2*np.pi*f1*x1) # Array of amplitudes at each time frame

plt.figure(figsize=(14,5))
plt.xlabel("time")
plt.ylabel("amplitude")
plt.title(f"Waveform of frequency {f1}")
plt.plot(x1[:pts],y1[:pts]) # Plot a small zoomed-in section of the sine wave
plt.show()

ipd.display(ipd.Audio(y1,rate=sr1))

d2 = 5 # Duration in seconds
sr2 = 22050 # Sample rate
f2 = 400 # Frequency
pts = 200 # number of array elements to plot

x2 = np.linspace(0,d2, sr2*d2) # Array of length equal to number of samples in 5 second audio
y2 = np.sin(2*np.pi*f2*x2) # Array of amplitudes at each time frame

plt.figure(figsize=(14,5))
plt.xlabel("time")
plt.ylabel("amplitude")
plt.title(f"Waveform of frequency {f2}")
plt.plot(x2[:pts],y2[:pts]) # Plot a small zoomed-in section of the sine wave
plt.show()
ipd.display(ipd.Audio(y2,rate=sr2))

# Composite wave
x3 = x1
y3 = y1 + y2
sr3 = sr1

plt.figure(figsize=(14,5))
plt.xlabel("time")
plt.ylabel("amplitude")
plt.title(f"Waveform of frequency {f1} and {f2} combined")
plt.plot(x3[:pts],y3[:pts]) # Plot a small zoomed-in section of the sine wave
plt.show()
ipd.display(ipd.Audio(y3,rate=sr3))

D = librosa.stft(y3) # Perform stft
print(D[:10])
S_db = librosa.amplitude_to_db(np.abs(D), ref=np.max) # Convert to logarithmic scale of dB
print(S_db[:10])
plt.figure(figsize=(14,5))
librosa.display.specshow(S_db, sr=sr3,x_axis="time", y_axis="hz")
plt.colorbar(format='%+0.2f dB')
plt.title("Spectogram")
plt.show()




```
